# 05 - 1D U-Net++ Architecture Experiments

Instantiates Model A (Baseline), Model B (Dilated), Model C (MHSA), and Model D (Enhanced), comparing parameters.


```python
import sys
from pathlib import Path
import torch
sys.path.insert(0, str(Path.cwd().parent / 'src'))

from earthquake_detection.models.unetplusplus_1d import UNetPlusPlus1D

models = {
    'Model A (Baseline)': UNetPlusPlus1D(in_channels=3, base_channels=16, depth=4),
    'Model B (Dilated)': UNetPlusPlus1D(in_channels=3, base_channels=16, depth=4, dilated_convolutions=True),
    'Model C (Attention)': UNetPlusPlus1D(in_channels=3, base_channels=16, depth=4, attention=True),
    'Model D (Enhanced)': UNetPlusPlus1D(in_channels=3, base_channels=16, depth=4, dilated_convolutions=True, attention=True, deep_supervision=True),
}

for name, m in models.items():
    params = sum(p.numel() for p in m.parameters() if p.requires_grad)
    x = torch.randn(2, 3, 1000)
    out = m(x)
    shape_str = f'{len(out)} heads' if isinstance(out, list) else str(out.shape)
    print(f'{name:<22}: {params:,} parameters | Output: {shape_str}')
```
