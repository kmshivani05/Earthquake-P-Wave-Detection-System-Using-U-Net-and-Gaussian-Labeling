# 06 - Phase Arrival Error Analysis & Benchmarks

Evaluates arrival residuals, calculates MAE and F1 metrics, and benchmarks noise robustness across SNRs.


```python
import sys
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
sys.path.insert(0, str(Path.cwd().parent / 'src'))

from earthquake_detection.evaluation.metrics import calculate_metrics
from earthquake_detection.visualization.plots import plot_error_distribution

# Synthetic ground truths vs predicted arrivals
gt = [900 + i for i in range(50)]
pred = [g + int(np.random.normal(0, 8)) for g in gt]

metrics = calculate_metrics(gt, pred, timing_tolerance_samples=20, sampling_rate_hz=100.0)
print(metrics.summary())

errors = [p - g for g, p in zip(gt, pred)]
fig = plot_error_distribution(errors, tolerance_samples=20, sampling_rate=100.0)
plt.show()
```
