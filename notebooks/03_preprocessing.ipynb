# 03 - Seismic Signal Preprocessing

Demonstrates 4th-order Butterworth bandpass filtering, per-channel normalization, and baseline detrending.


```python
import sys
from pathlib import Path
import h5py
import matplotlib.pyplot as plt
sys.path.insert(0, str(Path.cwd().parent / 'src'))

from earthquake_detection.preprocessing.filtering import apply_bandpass_filter
from earthquake_detection.preprocessing.normalization import normalize_waveform

with h5py.File('../STEAD/test.hdf5', 'r') as f:
    raw_trace = f['traces'][0].T  # (3, 6000)

filtered = apply_bandpass_filter(raw_trace, lowcut=0.5, highcut=10.0, sampling_rate=100.0, zero_phase=True)
normalized = normalize_waveform(filtered, method='per_channel')

plt.figure(figsize=(12, 3), dpi=150)
plt.plot(normalized[0], label='Normalized Filtered Z-Channel', color='#1f77b4', lw=0.8)
plt.title('Preprocessed Waveform (0.5-10 Hz Bandpass + Per-Channel Standardization)')
plt.legend()
plt.show()
```
