# 04 - Continuous Gaussian Target Generation

Explores Gaussian target formulation, standard deviation sigma tuning, and boundary condition handling.


```python
import sys
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
sys.path.insert(0, str(Path.cwd().parent / 'src'))

from earthquake_detection.data.gaussian_labels import generate_gaussian_target

T = 1000
t_p = 400

plt.figure(figsize=(10, 4), dpi=150)
for s in [10, 20, 30, 40]:
    target = generate_gaussian_target(T, t_p, sigma=s)
    plt.plot(target, label=f'sigma = {s} samples')

plt.axvline(t_p, color='red', linestyle='--', label=f'True Arrival (t={t_p})')
plt.title('Gaussian Target Distribution across Varying Widths (sigma)')
plt.xlabel('Time (Samples)')
plt.ylabel('Target Score')
plt.legend()
plt.show()
```
